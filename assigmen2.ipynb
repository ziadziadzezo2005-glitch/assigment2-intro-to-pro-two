```python
#question 1:
def strings_longer_than(strings, k):
    return [text for text in strings if len(text) > k]
print(strings_longer_than(["apple", "pie", "banana"], 3))
```

```python
#question 2:
import math
def area(shape, *dimensions):
    shape = shape.lower()

    if shape == "circle" and len(dimensions) == 1:
        radius = dimensions[0]
        return f"Circle Area: {math.pi * radius ** 2:.1f}"

    if shape == "rectangle" and len(dimensions) == 2:
        length, width = dimensions
        return f"Rectangle Area: {length * width}"

    raise ValueError(
        "Use area('circle', radius) or area('rectangle', length, width)."
    )


print(area("circle", 5))    
print(area("rectangle", 4, 6))
```

```python
#question 3:
def split_email(email):
    username, separator, domain = email.partition("@")
    if not separator or not username or not domain:
        raise ValueError("Enter a valid email address.")
    return username, domain


username, domain = split_email("support@google.com")
print(f"Username: {username}, Domain: {domain}")
```

```python
#question 4:
def sum_numbers(items):
    
    total = 0
    for item in items:
        
        if isinstance(item, (int, float)) and not isinstance(item, bool):
            total += item
    return total
 
 
if __name__ == "__main__":
    data = [10, "apple", 20, True, 5]
    print("Total Sum:", sum_numbers(data))
 
```

```python
#question 5:
class Restaurant:
    def __init__(self):
        self.menu_items = {}        
        self.book_table = {}        
        self.customer_orders = {}   
 
    def add_item_to_menu(self, item, price):
        self.menu_items[item] = price
        print(f"Added to menu: {item} - ${price}")
 
    def book_tables(self, table_number, customer_name):
        if table_number in self.book_table:
            print(f"Table {table_number} is already booked by {self.book_table[table_number]}.")
        else:
            self.book_table[table_number] = customer_name
            print(f"Table {table_number} booked for {customer_name}.")
 
    def customer_order(self, customer_name, item):
        if item not in self.menu_items:
            print(f"Sorry, '{item}' is not on the menu.")
            return
        self.customer_orders.setdefault(customer_name, []).append(item)
        print(f"{customer_name} ordered {item}.")
 
    def print_menu(self):
        print("\n--- Menu ---")
        for item, price in self.menu_items.items():
            print(f"{item}: ${price}")
 
    def print_reservations(self):
        print("\n--- Table Reservations ---")
        for table, name in self.book_table.items():
            print(f"Table {table}: {name}")
 
    def print_orders(self):
        print("\n--- Customer Orders ---")
        for name, items in self.customer_orders.items():
            total = sum(self.menu_items[i] for i in items)
            print(f"{name}: {', '.join(items)} (Total: ${total})")
 
 
if __name__ == "__main__":
    r = Restaurant()
 
    
    r.add_item_to_menu("Burger", 8)
    r.add_item_to_menu("Pizza", 12)
    r.add_item_to_menu("Pasta", 10)
    r.add_item_to_menu("Soda", 2)
 
    
    r.book_tables(1, "Ahmed")
    r.book_tables(2, "Sara")
    r.book_tables(1, "Omar")   # already booked
 
   
    r.customer_order("Ahmed", "Burger")
    r.customer_order("Ahmed", "Soda")
    r.customer_order("Sara", "Pizza")
    r.customer_order("Sara", "Sushi")   # not on menu
 
    
    r.print_menu()
    r.print_reservations()
    r.print_orders()
 
```

```python
#question 6:
class Customer:
    def __init__(self, name, age):
        self.__name = name
        self.__age = age
 
    def get_name(self):
        return self.__name
 
    def set_name(self, name):
        self.__name = name
 
    def get_age(self):
        return self.__age
 
    def set_age(self, age):
        self.__age = age
 
 
class Bike:
    def __init__(self, name, price):
        self.__name = name
        self.__price = price
 
    def get_name(self):
        return self.__name
 
    def set_name(self, name):
        self.__name = name
 
    def get_price(self):
        return self.__price
 
    def set_price(self, price):
        self.__price = price
 
 
class MainBikeRental:
    RATE = 40 
 
    def __init__(self):
        self.__bikes = []     
        self.__rentals = []   
 
    def get_bikes(self):
        return self.__bikes
 
    def set_bikes(self, bikes):
        self.__bikes = bikes
 
    def get_rentals(self):
        return self.__rentals
 
    def set_rentals(self, rentals):
        self.__rentals = rentals
 
    def addBike(self, bike):
        self.__bikes.append(bike)
 
    def totalCost(self):
        return sum(b.get_price() for b in self.__bikes)
 
    def requestBike(self, customer, num_of_bikes):
        if customer.get_age() <= 6:
            print(f"Sorry {customer.get_name()}, customers must be over 6 years old.")
            return False
        if num_of_bikes <= 0:
            print("Please request at least one bike.")
            return False
        if num_of_bikes > len(self.__bikes):
            print("Wait 10 mins")
            return False
 
        # Deduct bikes from stock and record the rental
        rented = [self.__bikes.pop() for _ in range(num_of_bikes)]
        self.__rentals.append({"customer": customer, "bikes": rented})
        print(f"{customer.get_name()} rented {num_of_bikes} bike(s). "
              f"Stock left: {len(self.__bikes)}")
        return True
 
    def returnBike(self, customer, rentalTime, num_of_bikes):
        for rental in self.__rentals:
            if rental["customer"] is customer and len(rental["bikes"]) >= num_of_bikes:
                returned = [rental["bikes"].pop() for _ in range(num_of_bikes)]
                self.__bikes.extend(returned)       # back into stock
                if not rental["bikes"]:
                    self.__rentals.remove(rental)
                bill = rentalTime * self.RATE
                print(f"{customer.get_name()} returned {num_of_bikes} bike(s). "
                      f"Total bill: {bill}")
                return bill
        print("No matching rental found for this customer.")
        return 0
 
 
if __name__ == "__main__":
    shop = MainBikeRental()
    shop.addBike(Bike("Mountain Bike", 300))
    shop.addBike(Bike("Road Bike", 250))
    shop.addBike(Bike("City Bike", 200))
    print("Total cost of bikes in store:", shop.totalCost())
 
    ali = Customer("Ali", 25)
    kid = Customer("Tim", 5)
 
    shop.requestBike(kid, 1)     
    shop.requestBike(ali, 5)     
    shop.requestBike(ali, 2)     
    print("Total cost of bikes in store:", shop.totalCost())
    shop.returnBike(ali, 3, 2)   
    print("Bikes in stock:", len(shop.get_bikes()))
```
